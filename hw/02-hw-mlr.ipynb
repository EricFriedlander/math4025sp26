```python
import numpy as np
import pandas as pd
from ISLP import load_data
Carseats = load_data('Carseats')
```

```python
y = Carseats['Sales']
x = Carseats['Price']
learning_rate = 10**(-6)
iterations = 100000
n = len(y)
# Initialize parameters
m = 0
b = 13

for i in range(iterations):
    # 1. Make a prediction
    y_predicted = m * x + b
    
    # 2. Calculate gradients (derivatives)
    # Derivative with respect to m
    dm = -(2/n) * np.sum(x * (y - y_predicted))
    # Derivative with respect to b
    db = -(2/n) * np.sum(y - y_predicted)
    
    # 3. Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db
    
    # Optional: Print progress every 100 steps
    if i % 100 == 0:
        cost = (1/n) * sum([val**2 for val in (y - y_predicted)])
        print(f"Iteration {i}: m={m:.4f}, b={b:.4f}, cost={cost:.4f}")
    


print(f"\nFinal Result: y = {m:.2f}x + {b:.2f}")
```
