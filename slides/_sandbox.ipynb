```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from itables
```

```python
outlets = pd.read_csv("../data/outlets.csv")
outlets.head()
```

```python
from plotnine import *
```

```python
(
    ggplot(outlets, aes(x="population", y="profit"))
        + geom_point()
        + geom_smooth(method="lm", se=False)
)
```

```python
model = LinearRegression()

model.fit(X=outlets[['population']], y=outlets['profit'])
```

```python
model_fit
```

```python
new_cities = pd.DataFrame({"population": np.random.normal(7, 3, 100)})
new_cities.head()
```

```python
model.predict(new_cities[['population']])
```
